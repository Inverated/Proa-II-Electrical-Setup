```python
import numpy as np
import pandas as pd
import os
```

```python
MAX_ROWS_PER_FILE = 500000
```

```python
file_name = "data_20260608-153041.csv"
df = pd.read_csv(file_name)
```

```python
print(df.head())
print(df.shape)
```

```python
def split_df(df, max_rows=MAX_ROWS_PER_FILE):
    df_list = []
    for i in range(0, len(df), max_rows):
        df_list.append(df.iloc[i:i + max_rows])
    return df_list
```

```python
df_list = split_df(df)
```

```python
rows = 0
for df in df_list:
    print(df.shape)
    rows += df.shape[0]

print(rows)
print("Number of splits:", len(df_list))
```

```python
for i, df_part in enumerate(df_list):
    output_file_name = f"{file_name.split('.csv')[0]}_part_{i+1}.csv"
    df_part.to_csv(output_file_name, index=False)
    print(f"Saved {output_file_name} with shape {df_part.shape}")
print("All files saved successfully.")
```
