```python
import numpy as np
import pandas as pd
from pathlib import Path
```

```python
current_directory = Path.cwd().parent
discharge_directory = Path.joinpath(current_directory, "Data Start Point")
ocv_soc_dict_file_name = Path.joinpath(discharge_directory, "ocv_soc_dict.csv")
battery_state = Path.joinpath(discharge_directory, "battery_state.csv")
```

```python
state_df = pd.read_csv(battery_state)
state_df.tail()
```

```python
ocv_soc_dict = pd.read_csv(ocv_soc_dict_file_name)
ocv_soc_dict.tail()
```

```python
state_df['OCV'] = ocv_soc_dict['OCV']
```

```python
state_df.tail()
```
