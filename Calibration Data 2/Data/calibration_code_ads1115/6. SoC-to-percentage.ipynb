```python
import numpy as np
import pandas as pd
from pathlib import Path
import json
```

```python
current_directory = Path.cwd().parent
discharge_directory = Path.joinpath(current_directory, "Data Start Point")
battery_state_file = Path.joinpath(discharge_directory, "battery_state.csv")
```

```python
df = pd.read_csv(battery_state_file)
df.tail()
```

```python
if df['SoC'].max() > 1:
    df['SoC'] = df['SoC'] / 100
df.tail()
```

```python
df['SoC'] = df['SoC'].round(9)
```

```python
df.to_csv(battery_state_file, index=False)
```
